## 1. What is Google Colab?
**Google Colaboratory** (or **Colab** for short) is a free, cloud-based platform that allows you to write and execute Python code through the browser. It's built on top of Jupyter Notebook and provides free access to computing resources including GPUs and TPUs.

Key features of Google Colab include:
- No setup required — just open a browser and start coding
- Free access to GPUs and TPUs for accelerated computing
- Easy sharing and collaboration through Google Drive integration
- Pre-installed libraries for data science and machine learning

## 2. The Colab Interface
The Colab interface is where you'll work with your notebooks. Here are the main components:

- **Menu bar**: At the top of the page with options like File, Edit, View, Insert, Runtime, Tools, and Help

- **Toolbar**: Below the menu bar with buttons for common actions like adding cells, running cells, and connecting to a runtime

- **Sidebar**: On the left side of the screen with tabs for:
  - **Table of contents**: Automatically generated based on your notebook headings
  - **Files**: Access to files uploaded to the Colab environment
  - **Snippets**: Useful code examples you can insert into your notebook

To access the sidebar, click on any of the icons in the interface. This will expand or collapse the sidebar panel.

To access additional tools and settings, you can use the three-dot menu in the upper right corner of the interface.

## 3. Working with Colab Notebooks
A Colab notebook is an interactive document that combines code, text, images, and visualizations. If you're looking at this during the course, then you are reading this text from a Colab notebook right now. Most of the work we will do will be inside notebooks like this, so here we're going to cover some of the details.

<a class="anchor" id="cells"></a>

### Cells
A notebook is divided up into what are called *cells*, which are blocks that contain information like code, text, or images. Different types of cells in a notebook are used to indicate how the content of that cell should be treated, these are typically "Text" (Markdown for web formatting) or "Code" (text we'd want to be executed as Python code).

This particular cell is set to Text (Markdown). You can identify a cell's type in Colab by inspecting its contents:
 
 - **Text cell**: Has formatted text, images, links, and equations using Markdown syntax. When run, it renders the formatting into a readable webpage-like format
 - **Code cell**: Has a play button (▶) on the left side that you can click to run the code. By default, code cells contain executable Python code and display any output below the cell. You can also run shell commands by starting the line with ! (e.g. !ls to list files)



#### Running a cell
Cells can be "run", or executed. If the cell is a "Text" cell (Markdown), it will be rendered to look like a webpage when it is run. If it is a "Code" cell, it will run the Python code and display any output.

This next cell is a Code cell. If we select it, and then click the "play" icon to the left of the cell, or use the keyboard shortcut, it will execute the code and print the output.

```python
5 + 5
```

There are also handy shortcut keys we can use to run a cell. We still first need to select the cell we want to run, then we can use these shortcut keys:

Run a cell:
- Mac:     `CMD + RETURN` (or `CMD + ENTER`)
- Windows: `CTRL + ENTER`

Run a cell and move to the next cell:
- Mac:     `SHIFT + RETURN` (or `SHIFT + ENTER`)
- Windows: `SHIFT + ENTER`

#### Adding a cell
We can add, remove, or move around cells with various icons on the screen. In Colab, to add a cell below the current one, you can either:

- Click the "+ Code" or "+ Text" buttons in the toolbar at the top
- Click the "+ Code" or "+ Text" buttons that appear when you hover between cells
- Use the keyboard shortcut `Ctrl+M B` (Windows) or `Cmd+M B` (Mac)

Below this cell, create a Code cell, type in `4 + 6`, and then try running the cell with the shortcut keys appropriate for the computer you are using.

If you did the step above, you should now see a cell above this one that has `4 + 6` in it with the answer printed below it.

### Runtimes
In Colab, the "Runtime" (similar to Jupyter's "Kernel") is the computing environment that executes your code. Colab provides free access to different runtime types, including CPU, GPU, and TPU runtimes.

To check your current runtime status, look at the upper-right corner of the notebook. You'll see either:
- "Disconnected" if no runtime is connected
- "Connected" if a runtime is active

You can change your runtime type by clicking on "Runtime" in the top menu and selecting "Change runtime type". This allows you to choose:
- **Runtime type**: Python 3 (default)
- **Hardware accelerator**: None, GPU, or TPU

When you're not actively using a notebook, Colab will eventually disconnect the runtime to conserve resources. You'll need to reconnect when you want to continue working. Also note that any variables or data in memory will be lost when the runtime disconnects.

To connect to a runtime, click the "Connect" button in the upper right corner of the notebook.
